## 3.2: Piecewise Approximation of an Arithmetic Function

In classical computing, functions like trigonometric, hyperbolic, and fractional powers are fundamental, often approximated using basic arithmetic operations such as addition and multiplication through polynomial approximations. Similarly, such functions are critical in fault-tolerant quantum algorithms. For instance, the HHL algorithm uses an eigenvalue inversion block requiring an $\arcsin(1/x)$ function, and quantum option-pricing applications may involve nonlinear payoff functions. Specifically, the hyperbolic tangent $\tanh(x)$ appears in applications such as Gibbs state preparation in statistical mechanics or as an activation function in quantum machine learning, with growing use cases expected as the field advances.



Your goal is to create a quantum function that is an accurate approximation of $\tanh(x)$ in the domain $f(x):[0,1)\rightarrow[0,1)$.  If you need a refresher on the hyperbolic tangent function, see [this Wikipedia page](https://en.wikipedia.org/wiki/Hyperbolic_functions).

To accomplish this, you will need to construct a quantum model that performs the following operation:


$$
|x\rangle_n\rightarrow|x\rangle_n|\tanh(x)\rangle_m
$$

You are advised to use a **piecewise linear approximation** to the function, with **4 pieces**. The grid resolution is given by `PRECISION_INPUT`, where the output value resolution is given by `PRECISION_RESULT`. Because we are interested in values between $0$ and $1$, the input and output numbers will be interpreted as binary fixed-point values: `0.xxxxxx`, where the precision is the number of bits to the right of the binary point.

The following is some code that will help you create the piecewise linear approximation. This function uses the Numpy library to create a Chebyshev polynomial for a given function (`func`) in the range between `x0` and `x1`.  The second parameter to the `interpolation` call is 1, which means this is a linear approximation, and there are two coefficients returned.  Recall that a linear function is of the form: $$y = c_1 x + c_0,$$ where $c_1$ is the slope and $c_0$ is the y-offset. These are the two coefficients returned by the function.


````python
def approximate_piece(func, x0, x1):
    # use a chebyshev approximation:
    cheb_poly = Chebyshev.interpolate(lambda x: func(x+x0), 1, (0, x1-x0))
    # Convert Chebyshev polynomial to standard polynomial
    poly = cheb_poly.convert(kind=Polynomial)
    # Access coefficients of the standard polynomial
    coefficients = poly.coef
    return poly.coef[1], poly.coef[0]
````

The following is a function for evaluating the accuracy of the result, where the measure is the maximum absolute difference between the function value and the approximation, on all given input points. It will also plot the approximation along with the exact function.

````python
def evaluate_L_inf(results, evaluation_grid_precision=PRECISION_INPUT):
    parsed_counts = sorted(results.parsed_counts, key=lambda x: x.state["x"])
    res = {s.state['x']: s.state['y'] for s in parsed_counts}
    user_input_precision = len(results.output_qubits_map['x'])
    assert len(res) == 2 ** user_input_precision # verify all strings were sampled, also no superpositions
    domain = np.arange(0, 1, 1/ 2**evaluation_grid_precision)
    expected_y = np.tanh(domain)
    measured_y = []
    for input in domain:
        input_floored = int(input * (2**user_input_precision)) / (2**user_input_precision)
        measured_y.append(res[input_floored])

    plt.scatter([s.state['x'] for s in parsed_counts],
                [s.state['y'] for s in parsed_counts], label="measured")
    plt.plot(domain, expected_y, label="expected")
    plt.legend()
    plt.show()
    plt.interactive(False)
    l_inf = np.max(np.abs(expected_y - np.array(measured_y)))
    return l_inf
````
