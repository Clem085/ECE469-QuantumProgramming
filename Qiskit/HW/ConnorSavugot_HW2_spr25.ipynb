```python
def increment_4bit_qc(input_value: int, increment: int, display_mode: str = "bloch"):
    """
    Create a quantum circuit that adds a fixed 4-bit unsigned integer to a 4-qubit input.
    
    Parameters:
        input_value (int): The initial 4-bit unsigned integer.
        increment (int): The unsigned 4-bit integer value to add.
        display_mode (str): "bloch" to display Bloch sphere visualization,
                           "table" to display qubit states in table format.
    
    Returns:
        QuantumCircuit: The quantum circuit implementing the increment operation using mcx gates.
    """
    if not (0 <= input_value <= 15):
        raise ValueError("Input value must be a 4-bit unsigned integer (0 to 15).")
    if not (0 <= increment <= 15):
        raise ValueError("Increment must be a 4-bit unsigned integer (0 to 15).")
    
    qr = QuantumRegister(4, 'q')  
    qc = QuantumCircuit(qr) 
    
    # --- Set up the quantum input state to match the parameter input_value ---
    binary_input = f"{input_value:04b}"[::-1]  # Reverse for little-endian order
    initial_state = Statevector.from_int(input_value, 2**4)  # Represents the 4-qubit input state
    
    # print("TEST: Binary Input:",binary_input)

    for i in range(3,-1,-1):
        if binary_input[i] == '1':
            # print("TEST: X Gate Placed at Qubit:",2**i)
            qc.x(qr[i])
    
    # Add a visual barrier to separate initialization from computation
    qc.barrier()
    
    # Addition Operation
    # Convert increment value to 4-bit binary (LSB first for proper mapping)
    binary_str = f"{increment:04b}"[::-1]  # Reverse binary to match qubit indexing

    # Apply X gates for bits set to 1
    for i in range(3,-1,-1):
        if binary_str[i] == '1':  
            # Apply carry propagation using mcx (LSB -> MSB)
            for j in range(3,i,-1):  # Start carry from current bit up to MSB
                control_qubits = qr[i:j]  # All previous qubits control the next target #qr[:j]
                target_qubit = qr[j]  # Next bit to be modified
                qc.mcx(control_qubits, target_qubit)  # Apply carry propagation
            qc.x(qr[i])  # Flip bit directly

    
    # Simulate Circuit Final State
    final_state = Statevector(qc)
    final_binary_state = f"{(input_value + increment) % 16:04b}"[::-1]
    
    if display_mode == "table":
        print(f"Input Output Table incrementing {input_value} by {increment}:")
        latex_str = r"\begin{array}{|c|c|c|} \hline "
        latex_str += r"& \text{Initial Value} & \text{Final Value} \\ \hline "
        for i in range(3,-1,-1):
            initial_qubit = binary_input[i]
            final_qubit = final_binary_state[i]
            latex_str += fr"Qubit {i} & |{initial_qubit}\rangle & |{final_qubit}\rangle \\ \hline "
        latex_str += r"\end{array}"
        display(Math(latex_str))

    elif display_mode == "bloch":
        print(f"Input Qubit Bloch Diagram Incrementing {input_value} by {increment}:")
        display(plot_bloch_multivector(initial_state, reverse_bits=True))
        print(f"Output Qubit Bloch Diagram Incrementing {input_value} by {increment}:")
        display(plot_bloch_multivector(final_state, reverse_bits=True))
    else:
        # Useless, but included to remind me I have a third potential Value for display_mode Parameter
        # Random Rotation Circuits must be printed later, because it is actually created by combining two different quantum circuits
        print("",end="")
    return qc


def random_superposition_circuit():
    """
    Generates a quantum circuit where all 4 qubits are in a random superposition
    by applying rotations across multiple axes (X, Y, Z).
    
    Returns:
        QuantumCircuit: The quantum circuit with superposition initialization.
    """
    qr = QuantumRegister(4, 'q')
    qc = QuantumCircuit(qr)

    # Apply random rotations to create non-trivial superposition
    for i in range(4):
        theta_x = np.random.uniform(0, 2 * np.pi)  # Random angle for RX
        theta_y = np.random.uniform(0, 2 * np.pi)  # Random angle for RY
        theta_z = np.random.uniform(0, 2 * np.pi)  # Random angle for RZ

        qc.rx(theta_x, qr[i])  # Rotate around X-axis
        qc.ry(theta_y, qr[i])  # Rotate around Y-axis
        qc.rz(theta_z, qr[i])  # Rotate around Z-axis

    return qc



### Test Cases
# Demonstrate for input value 4 and increment value 1
input_value = 8
increment_value = 1
qc = increment_4bit_qc(input_value, increment_value, display_mode="bloch")
print(f"Circuit for incrementing {input_value} by {increment_value}:")
display(qc.draw('mpl'))
    
# Demonstrate for increment values 5, 6, and 10
increment_values = [5, 6, 10]
circuits = {}
for inc in increment_values:
    circuits[inc] = increment_4bit_qc(4, inc, display_mode="bloch")  # Use input value 4 as example
    print(f"Circuit for incrementing {4} by {inc}:")
    display(circuits[inc].draw('mpl'))

# Test ALL Classical Input States
for input_value2 in range(0,16):
    for inc2 in increment_values:
        print(f"Circuit for incrementing {input_value2} by {inc2}:")
        circuits[input_value2] = increment_4bit_qc(input_value2, inc2, display_mode="table")
        display(circuits[input_value2].draw('mpl'))

# Test 10 Random Superposition States
for test_num in range(10):
    print(f"Test {test_num + 1}: Incrementing a Random Superposition State")
    
    # Generate a random superposition circuit
    superposition_circuit = random_superposition_circuit()
    
    # Get initial state before applying the increment
    initial_state = Statevector(superposition_circuit)

    # Select a random increment value (from predefined cases)
    inc3 = np.random.choice([5, 6, 10])  
    
    # Create increment circuit
    increment_circuit = increment_4bit_qc(0, inc3, display_mode="random")
    
    # Combine superposition + increment circuits
    full_circuit = superposition_circuit.compose(increment_circuit)

    # Get final state after applying the increment operation
    final_state = Statevector(full_circuit)

    # Print Bloch Sphere full_circuit Input and Output
    print(f"Input Qubit Bloch Diagram Incrementing a Random Superposition by {inc3}:")
    display(plot_bloch_multivector(initial_state, reverse_bits=True))

    print(f"Output Qubit Bloch Diagram Incrementing a Random Superposition by {inc3}:")
    display(plot_bloch_multivector(final_state, reverse_bits=True))

    # Convert quantum state vector into LaTeX table format
    print(f"Input Output Table incrementing a Random Superposition by {inc3}:")

    latex_str = r"\begin{array}{|c|c|c|} \hline "
    latex_str += r"& \text{Initial Vector Position} & \text{Final Vector Position} \\ \hline "

    # Extract quantum state vector data
    initial_vector = initial_state.data
    final_vector = final_state.data

    # Extract first 4 qubit states
    for i in range(3, -1, -1):
        initial_val = f"{initial_vector[i].real:.3f} + {initial_vector[i].imag:.3f}i"
        final_val = f"{final_vector[i].real:.3f} + {final_vector[i].imag:.3f}i"
        latex_str += fr"Qubit {i} & {initial_val} & {final_val} \\ \hline "

    latex_str += r"\end{array}"
    display(Math(latex_str))

    # Print circuit visualization
    print(f"Incrementing a random superposition by {inc3}:")
    display(full_circuit.draw('mpl'))
```

Define a **function** that contructs a circuit to test whether a four-qubit input is equal to a specified $4$-bit value ($n$). The circuit will use five qubits: four for the register to test, and one for the result.

Draw the circuit returned by the function for $n = 6$. Show the result of running the circuit on an input state that is an *equal superposition* of all computational basis states.

```python
def equality_test_4bit(n: int):
    """
    Constructs a quantum circuit to check if a 4-qubit register equals a specified 4-bit value (n).
    
    Parameters:
        n (int): The 4-bit value to compare (0-15).
    
    Returns:
        QuantumCircuit: A circuit that sets an ancilla qubit to |1> if the input register equals n.
    """
    if not (0 <= n <= 15):
        raise ValueError("n must be a 4-bit integer (0 to 15).")
    
    # Create quantum registers
    qr = QuantumRegister(5, 'q')  # 4 qubits for input, 1 qubit for result
    cr = ClassicalRegister(1, 'c')  # Classical bit to store the result
    qc = QuantumCircuit(qr, cr)

    # Convert n to binary and apply X gates to match the target pattern
    binary_n = f"{n:04b}"[::-1]  # Reverse to match little-endian Qiskit indexing
    for i in range(4):
        if binary_n[i] == '0':  
            qc.x(qr[i])  # Flip input qubits where n has 0s

    # Multi-Controlled Toffoli (mcx) - If all four qubits match n, flip ancilla qubit
    qc.mcx(qr[:4], qr[4])  

    # Uncompute (restore original state)
    for i in range(4):
        if binary_n[i] == '0':  
            qc.x(qr[i])

    # Measure result qubit
    qc.measure(qr[4], cr[0])
    
    return qc

# Create a circuit for n = 6
n = 6
qc_test = equality_test_4bit(n)
print(f"Quantum circuit for checking if a 4-qubit register equals {n}:")
display(qc_test.draw('mpl'))


# Create a circuit where input is an equal superposition
qr = QuantumRegister(5, 'q')
cr = ClassicalRegister(1, 'c')
qc_super = QuantumCircuit(qr, cr)

# Put input qubits into equal superposition
qc_super.h(qr[:4])

# Apply equality test for n = 6
qc_super.compose(equality_test_4bit(6), inplace=True)

# Show circuit
print("Superposition-based Equality Test Circuit:")
display(qc_super.draw('mpl'))

# Run multiple test cases
test_values = [0, 3, 6, 9, 12, 15]  # Different 4-bit numbers
circuits = {}

for n in test_values:
    print(f"Quantum circuit for checking if a 4-qubit register equals {n}:")
    qc_test = equality_test_4bit(n)
    circuits[n] = qc_test  # Store circuit for later use
    display(qc_test.draw('mpl'))

# Superposition Test: Check all basis states at once
print("Superposition-based Equality Test Circuit:")
qr = QuantumRegister(5, 'q')
cr = ClassicalRegister(1, 'c')
qc_super = QuantumCircuit(qr, cr)

# Put input qubits into equal superposition
qc_super.h(qr[:4])

# Apply equality test for n = 6
qc_super.compose(equality_test_4bit(6), inplace=True)

# Show circuit
display(qc_super.draw('mpl'))


# Run multiple test cases
test_values = [0, 3, 6, 9, 12, 15]  # Different 4-bit numbers
circuits = {}

for n in test_values:
    print(f"Quantum circuit for checking if a 4-qubit register equals {n}:")
    qc_test = equality_test_4bit(n)
    circuits[n] = qc_test  # Store circuit for later use
    display(qc_test.draw('mpl'))

# Superposition Test: Check all basis states at once
print("Superposition-based Equality Test Circuit:")
qr = QuantumRegister(5, 'q')
cr = ClassicalRegister(1, 'c')
qc_super = QuantumCircuit(qr, cr)

# Put input qubits into equal superposition
qc_super.h(qr[:4])

# Apply equality test for n = 6
qc_super.compose(equality_test_4bit(6), inplace=True)

# Show circuit
display(qc_super.draw('mpl'))

```

#### Exercise 2.3 (25%)

Using the functions from 2.1 and 2.2 as building blocks, define a **function** that builds a circuit that has two four-qubit quantum registers -- we'll call them $a$ and $b$. The function should include a 4-bit unsigned integer ($n$), and it should increment $b$ by $n$ if and only if $a = n$.  The arguments to your function will be two `QuantumRegister` objects and the integer $n$.

The extra qubit used for the equality test (2.2) should not be included in the parameters of your function. That qubit will be created and used by your function, but will not be provided by the caller of your function. In other words, this is an *ancilla* qubit that is only used for a temporary purpose.

Demonstrate the circuit.

```python
def conditional_increment(a: QuantumRegister, b: QuantumRegister, n: int):
    """
    Constructs a circuit that increments the 4-qubit register b by n if and only if register a equals n.
    
    Parameters:
        a (QuantumRegister): The 4-qubit register to check.
        b (QuantumRegister): The 4-qubit register to increment.
        n (int): The unsigned 4-bit integer value.
    
    Returns:
        QuantumCircuit: The quantum circuit implementing the conditional increment.
    """
    if not (0 <= n <= 15):
        raise ValueError("n must be a 4-bit unsigned integer (0 to 15).")
    
    anc = QuantumRegister(1, 'anc')  # Ancilla qubit for equality test
    cr = ClassicalRegister(4, 'c')  # Classical register for measurement
    qc = QuantumCircuit(a, b, anc, cr)
    
    # Apply equality test on register a
    equality_circuit = equality_test_4bit(n)
    equality_circuit.remove_final_measurements()  # Remove classical measurement from equality test
    qc.compose(equality_circuit, qubits=a[:] + [anc[0]], inplace=True) # Use Compose instead of Append the prevent circuit logic from being abstracted
    
    # Increment b if a = n (controlled on ancilla qubit)
    increment_circuit = increment_4bit_qc(0, n)
    increment_circuit.remove_final_measurements()  # Ensure increment circuit is unitary
    
    # **New Fix: Remove barriers from the increment circuit**
    increment_circuit.data = [gate for gate in increment_circuit.data if gate.operation.name != "barrier"]

    increment_gate = increment_circuit.to_gate().control(1)
    
    qc.compose(increment_gate, qubits=[anc[0]] + b[:], inplace=True)

    
    # Reset ancilla state to 0
    qc.compose(equality_circuit, qubits=a[:] + [anc[0]], inplace=True) # Use Compose instead of Append the prevent circuit logic from being abstracted
    
    # Measure register b
    qc.measure(b, cr)
    
    return qc

# Demonstrate circuit
qa = QuantumRegister(4, 'a')
qb = QuantumRegister(4, 'b')
qc_demo = conditional_increment(qa, qb, 6)
print("Quantum circuit for conditional increment when a = 6:")
display(qc_demo.draw('mpl'))


# Test Case 1: a = n (expected increment)
qa1 = QuantumRegister(4, 'a')
qb1 = QuantumRegister(4, 'b')
qc_test1 = conditional_increment(qa1, qb1, 6)
print("Test Case 1: a = 6, should increment b by 6")
display(qc_test1.draw('mpl'))

# Test Case 2: a ≠ n (expected no increment)
qa2 = QuantumRegister(4, 'a')
qb2 = QuantumRegister(4, 'b')
qc_test2 = conditional_increment(qa2, qb2, 3)  # Set n = 3, but initialize a to a different value
print("Test Case 2: a ≠ n, b should remain unchanged")
display(qc_test2.draw('mpl'))

# Test Case 3: Edge case, n = 0 (b should remain the same)
qa3 = QuantumRegister(4, 'a')
qb3 = QuantumRegister(4, 'b')
qc_test3 = conditional_increment(qa3, qb3, 0)
print("Test Case 3: n = 0, b should not change")
display(qc_test3.draw('mpl'))

# Test Case 4: Edge case, n = 15 (maximum increment)
qa4 = QuantumRegister(4, 'a')
qb4 = QuantumRegister(4, 'b')
qc_test4 = conditional_increment(qa4, qb4, 15)
print("Test Case 4: n = 15, b should increment by 15")
display(qc_test4.draw('mpl'))



```

Define a **function** that creates a quantum circuit that creates a 4-qubit equal superposition of computational basis states and then selectively changes the phase of a designated set of basis states. The function has one parameter: a list of integer values that specify the computational bases that will be phase-shifted.  (You may assume that the values are 4-bit integers (0-15).)

For this assignment, the desired phase shift is $90^\circ$ (or $\pi/2$).

For example, suppose the argument is `[2, 9]`.  The output is an equal superposition of $|0\rangle$ through $|15\rangle$ (hint: Hadamard gates!) and states $|2\rangle$ and $|9\rangle$ will be rotated by $\pi/2$.

Draw the circuit and demonstrate that it works correctly for at least two different "interesting" lists. (Consider how to clearly display that the circuit performs correctly.)

You may define additional functions, if desired, and you may use functions that were defined in previous exercises.

```python
def apply_manual_controlled_phase(qc, controls, ancilla, angle):
    """
    Apply a controlled phase shift of `angle` using only mcx (multi-controlled Toffoli) and single-qubit phase gates.
    """
    if len(controls) == 1:
        qc.p(angle, controls[0])  # Apply a single-qubit phase shift
    else:
        qc.mcx(controls, ancilla)  # Multi-controlled Toffoli to set ancilla
        qc.p(angle, ancilla)  # Apply phase shift to ancilla
        qc.mcx(controls, ancilla)  # Uncompute ancilla

def phase_shift_superposition(phase_states: list):
    """
    Creates a quantum circuit that generates a 4-qubit equal superposition
    and applies a phase shift of π/2 (90 degrees) to specified computational basis states.

    Parameters:
        phase_states (list): List of integers (0-15) representing basis states to phase shift.

    Returns:
        QuantumCircuit: The quantum circuit implementing the operation.
    """
    if not all(0 <= state <= 15 for state in phase_states):
        raise ValueError("All states must be 4-bit integers (0 to 15).")
    
    qr = QuantumRegister(4, 'q')  # 4 qubits for superposition
    ancilla = QuantumRegister(1, 'anc')  # Single ancilla qubit for phase shifting
    cr = ClassicalRegister(4, 'c')  # Classical register (if needed)
    qc = QuantumCircuit(qr, ancilla, cr)
    
    # Step 1: Create Equal Superposition
    qc.h(qr)  # Applies Hadamard gate to each qubit
    
    # Step 2: Apply Phase Shift of π/2 to Specific Basis States using mcx
    for state in phase_states:
        binary_state = f"{state:04b}"[::-1]  # Reverse for little-endian ordering
        controls = [qr[i] for i in range(4) if binary_state[i] == '1']
        
        # Ensure phase shift is applied without using `cp()`
        if len(controls) == 1:
            qc.p(np.pi / 2, controls[0])  # Direct phase shift
        elif len(controls) > 1:
            apply_manual_controlled_phase(qc, controls, ancilla[0], np.pi / 2)  # Use ancilla

    return qc

# Example Demonstrations
phase_lists = [[2, 9], [3, 12], list(range(16))]  # Multiple test cases
circuits = {}
for phase_list in phase_lists:
    print(f"Quantum circuit with phase shift on states: {phase_list}")
    qc = phase_shift_superposition(phase_list)
    display(qc.draw('mpl'))


# Test Cases for conditional_increment
test_values = [0, 3, 6, 7, 12, 15]  # Various test cases including edge values
circuits = {}

for test_n in test_values:
    print(f"Testing conditional increment for a = {test_n}")
    qa = QuantumRegister(4, 'a')  # 4-bit register a
    qb = QuantumRegister(4, 'b')  # 4-bit register b
    qc_test = conditional_increment(qa, qb, test_n)
    display(qc_test.draw('mpl'))  # Draw the quantum circuit

```
